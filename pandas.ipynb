## Selection by label
- See more in Selection by Label using DataFrame.loc() or DataFrame.at().
- Selecting a row matching a label:

```python
df.loc[dates[0]]
```

```python
# Selecting all rows (:) with a select column labels:
df.loc[:, ["A", "B"]]
```

```python
# For label slicing, both endpoints are included:
df.loc["20130102":"20130104", ["A", "B"]] # df.loc[row_labels, column_labels]
# row=20130102 to 20130104, columns A to B
```

```python
# Selecting a single row and column label returns a scalar:
df.loc[dates[0], "A"]
```

```python
# For getting fast access to a scalar (equivalent to the prior method):
df.at[dates[0], "A"]  # df.at is optimized for getting and setting a single value in the DataFrame,
#  so it's faster than df.loc for this specific use case.
```

# Selection by position
- Selection by Position using DataFrame.iloc() or DataFrame.iat().
- The position of the passed integers:

```python
df.iloc[3] # selects the fourth row of the DataFrame (since indexing starts at 0).
```

The fundamental difference between **.loc** and **.iloc** comes down to how they look up data:\
 **.loc** is label-based (it looks for names), while **.iloc** is integer-based (it looks for numeric index positions).

```python
# Integer slices acts similar to NumPy/Python:
df.iloc[3:5, 0:2]
```

```python
# Lists of integer position locations:
df.iloc[[1, 2, 4], [0, 2]]
```

```python
# For slicing rows explicitly:
df.iloc[1:3, :] 
# selects rows 1 and 2 (since the end index is exclusive) and all columns (:) of the DataFrame.
```

```python
# For slicing columns explicitly:
df.iloc[:, 1:3]
# selects all rows (:) and columns 1 and 2 (since the end index is exclusive) of the DataFrame.
```

```python
# For getting a value explicitly:
df.iloc[1, 1]
```

```python
# For getting fast access to a scalar (equivalent to the prior method):
df.iat[1, 1]
```

The fundamental difference between **.iat** and **.iloc** is speed and scope: **.iat** is a specialized tool designed to extract or change a single, individual cell as fast as possible, while **.iloc** is a general tool that can select multiple rows, columns, or entire slices of data.Both use integer-based positions (0, 1, 2...) to look up data, but they serve different optimization needs.

# Merge
## Concat
- pandas provides various facilities for easily combining together Series and DataFrame objects with various kinds of set logic for the indexes and relational algebra functionality in the case of join / merge-type operations.

```python
# Concatenating pandas objects together row-wise with concat():
df = pd.DataFrame(np.random.randn(10, 4))
df
```

```python
# break it into pieces
pieces = [df[:3], df[3:7], df[7:]]
```

```python
pieces
```

```python
pd.concat(pieces)
```

# Join

```python
left = pd.DataFrame({"key": ["foo", "foo"], "lval": [1, 2]})
left
```

```python
right = pd.DataFrame({"key": ["foo", "foo"], "rval": [4, 5]})
right
```

```python
pd.merge(left, right, on="key")
```

# Grouping
By “group by” we are referring to a process involving one or more of the following steps:

- Splitting the data into groups based on some criteria
- Applying a function to each group independently
- Combining the results into a data structure
